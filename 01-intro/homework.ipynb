QUESTION 7

```python
asia_df = df.loc[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
asia_df
```

```python
asia_df = asia_df.head(7)
asia_df
```

```python
X = asia_df.to_numpy()
```

```python
XTX = np.dot(X.T, X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
XTX_inv_multiplied_XT = np.dot(XTX_inv, X.T)
XTX_inv_multiplied_XT
```

```python
w = np.dot(XTX_inv_multiplied_XT, y)
w
```
